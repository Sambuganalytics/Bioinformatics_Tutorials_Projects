```python
# importing libraries
import Bio
from Bio.Seq import Seq
from collections import Counter
```

```python
# Let's write a cool function to verify validity of sequece first
def Seq_validity(sequence):
    allowed_nucs = ['A','G','C','T','U']
    sequence = sequence.upper()
    for ind,nuc in enumerate(sequence):
        if nuc not in allowed_nucs:
            return print(f"Your sequence has a wrong nucleotide '{nuc}' at index: {ind}")
    return print(f"Your sequence is okay :{sequence}")
```

```python
# Let's create our first sequece object to test our functoin
seq1 = Seq("GATCGATGGGCCTATATAGGATCGAAAATCSC")
Seq_validity(seq1)
```

```python
# We can use this do count and determine the GC fraction in our sequence
print(((my_seq.count('G') + my_seq.count('C'))/len(my_seq))*100) # that is 46.87 %
```

#### **2.0 Transcription and Translation**
---
- This part of the tutorial looks at DNA transcription and translation
- In biopython and most bioinformatic programs, we use the coding DNA sequence directly
- To simulate real biological event, the DNA_template strand can be synthesized using this rule:
- We need to flip the coding-strand to get the 3'-5' direction and then complement it.
- This way, we create the DNA-template i.e the antisense strand
- We can now create a complementary strand from this template in the 5'-3' direction for transcription
- **NB: The transcribe and translate methods have arguments: inplace for transcribe, table, etc for translate**

```python
# let's create a Seq object called coding sequence
coding_dna = Seq("ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG") # this is in 5'-3' direction

# in real biological life the DNA template or the antisense strand is in 3'- 5' direction
template_dna = coding_dna.reverse_complement()
```

```python
# Let's have a look at our sequences : added some somall directional formmating 
pairing_bars =  ''.join(['|' for i in range(len(coding_dna))])
print("5'",coding_dna,"3'")
print("  ", pairing_bars)
print("3'",template_dna,"5'")
```

##### **NB: This is how it appears biologically, they are however antiparalell.Textbook orientation is for easy visual understanding**
---
- No that the template if used in transcribing, has to be in the 5'- 3' direction and complemented
- The results is a complementary strand moving in the same direction and sequence as the coding strand
- That is why in biopython, we work with the coding strand directly: 
- We only need to change T for U , as in RNA complementarity rule

```python
# Let's now transcribe the coding and template dnas
transcribe_coding = coding_dna.transcribe()
print(transcribe_coding)
```

```python
# now for us to transcribe the dna_template, we need to get it's complementary
# Note that it is in 3'-5' and must be reversed and complemented
temp_complement = template_dna.reverse_complement()
print(temp_complement.transcribe())
```

###### **You see how that works: Both transcribed sequence are the same**

```python
# let's say you have an mRNA and you want to send it back to the coding template
mRNA = coding_dna.transcribe()
# use back.transcribe
back_coding = mRNA.back_transcribe()
print('mRNA:',mRNA)
print('Codng_dna :',coding_dna) 
print('back_trans_dna:',back_coding)
```

##### **2.1 Translation :**

```python
# Let's translate this mRNA we created 
protein1 = mRNA.translate() 
protein1
```

```python
# You can translate directly from the coding strand
protein1a = coding_dna.translate()
protein1a
```

You should notice in the above protein sequences that in addition to the end stop character, there is an internal stop as well. This was a deliberate choice of example, as it gives an excuse to talk about some optional arguments, including different translation tables (Genetic Codes).

The translation tables available in Biopython are based on those from the NCBI (see the next section of this tutorial). By default, translation will use the standard genetic code (NCBI table id 1). Suppose we are dealing with a mitochondrial sequence. We need to tell the translation function to use the relevant genetic code instead:

```python
# let's specify which table we are using: we are using a "Vertebrate Mitochondria" in this case
protein1b = mRNA.translate(table= "Vertebrate Mitochondrial")
protein1b
```

```python
# You can also specify the table using the NCBI table number on the tables
# let's use table 2; which is the "Vertebrate Mitochondria" translation table
protein1b = mRNA.translate(table=2)
protein1b
```

##### **NB: You see the table number 2 produces the same thing as using the full table name in words**
---
- You must be asking where comes this asteriks?
- Those asteriks represents where there are stop codons according to the gene table and the arguments to translate
- If you use, to_stop = True, it stops translating the mRNA at the first stop codon and the resulting protein has no asterik
- If you don't specify,the default is set to false and it is read and replaced with asterik

```python
# Now let's use the to_stop argument and see what happens to the mRNA been translated
protein1c = mRNA.translate(table=2, to_stop = True)
print('Include stop codon position :',protein1b)
print('Exclude stop codon ends tran :', protein1c)
```

```python
# You can specify which symbol to use for stop codon:
protein_ss = mRNA.translate(table=2, stop_symbol= '$') # let's put that sign where there is a stop codon
protein_ss
```

##### **NB:**
###### **In general, given a complete CDS, the default translate method will do what you want (perhaps with the to_stop option). However, what if your sequence uses a non-standard start codon? This happens a lot in bacteria – for example the gene yaaX in E. coli K12:**
###### **In the bacterial genetic code GTG is a valid start codon, and while it does normally encode Valine, if used as a start codon it should be translated as methionine. This happens if you tell Biopython your sequence is a complete CDS:**

```python
# Let's illustrate that with an example :
# Example of a complete coding sequecne
gene_bacteria = Seq(
    "GTGAAAAAGATGCAATCTATCGTACTCGCACTTTCCCTGGTTCTGGTCGCTCCCATGGCA"
    "GCACAGGCTGCGGAAATTACGTTAGTCCCGTCAGTAAAATTACAGATAGGCGATCGTGAT"
    "AATCGTGGCTATTACTGGGATGGAGGTCACTGGCGCGACCACGGCTGGTGGAAACAACAT"
    "TATGAATGGCGAGGCAATCGCTGGCACCTACACGGACCGCCGCCACCGCCGCGCCACCAT"
    "AAGAAAGCTCCTCATGATCATCACGGCGGTCATGGTCCAGGCAAACATCACCGCTAA")
```

```python
# Let's translate this: This sequece doesn't have a standard start codon(AGT)
# So GTG will normally code for Valine.
protein_bac1 = gene_bacteria.translate(table='Bacterial', to_stop= True)
protein_bac1
```

```python
# Now you see that in protein_bac1. the start codon is Valine
# We can want it to be methionine so we specify cds to true
protein_bac2 = gene_bacteria.translate(table='Bacterial', to_stop= True, cds= True)
print(protein_bac2)
```

##### **Now you see it converts the stars codone to methionine**

#### **6.0 MutableSeq objects**
---
- Normal string and Seq() objects are immutable:
- It means once written you cannot make any changes to it again.
- However, we can use mutableseq() to make the sequece mutable

```python
# let's use this aligned part of a partially defined human gene sequence
human_seq = Seq('TTGAAAACCTGAATGTGAGAGTCAGTCAAGGATAGT', length= 36)
```

```python
# if i try to write to it, or remove a base, it will throw an error 
#human_seq[0] = 'G'
# that won't work
```

```python
# Now let's make it mutable 
from Bio.Seq import MutableSeq
```

```python
# Now let's create a MutableSeq() object
human_mut = MutableSeq(human_seq)
# Let's change the first nucleotide to A
human_mut[0] = 'A'
```

```python
# You can see that now i have changed the first nucleotide
print('unmuttable->', human_seq)
print('Mutable ->', human_mut)
```
