#### **1.0 SeqIO Module**
---
- This notebook explores who to use biopython SeqIO module in parsing files containing biological information
- Using SeqIO returns an iterable which is made of SeqRecord objects

```python
# importing libraries
import Bio
from Bio.Seq import Seq
from Bio.SeqFeature import SeqFeature, FeatureLocation
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO
```

##### **1.1 Reading a Sequence file with SeqIO.parse()**
---

```python
# let's parse our first record file 
records = SeqIO.parse(r"D:\CLINISTATS\Gene_files\f002.fasta", 'fasta')
```

```python
for record in records:
    print('ID :', record.id, ':', 'Description',record.description)
```

##### **1.2 Uing a list to get records**
---
- Sometitmes you may want to have the records in a list for ordered accessing 
- The normal list function in python can be used 

```python
# say if we wanted to extract onl y the ids of these records, we would have used a list comprehension
id = [record.id for record in SeqIO.parse(r"D:\CLINISTATS\Gene_files\f002.fasta", 'fasta')]
id
```

```python
# you could equally use the list function to create it 
list(record.id for record in SeqIO.parse(r"D:\CLINISTATS\Gene_files\f002.fasta", 'fasta'))
```

```python
# say we want to grab the last record 
records_ls = list(SeqIO.parse(r"D:\CLINISTATS\Gene_files\f002.fasta", 'fasta'))

# length of records
print('There are', len(records_ls),'records in the file')

# spaces
print("********************************")

# the first record 
print(records_ls[0])
```

```python
# the last recond
print(records_ls[-1])
```

```python
# sometimes you need only the first record in iterable, you can use next() to get the record
first_record = next(SeqIO.parse(r"D:\CLINISTATS\Gene_files\f002.fasta", 'fasta'))
print(first_record) # this is the first record : compare it to record_ls[0]
```
